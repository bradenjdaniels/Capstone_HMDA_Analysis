```python
unique_leis = hmda["lei"].unique()
len(unique_leis)
#unique_leis = unique_leis[:200] #Comment in prod, used for testing
```

```python
lei_dict = {}

for lei in unique_leis:
    if len(lei_dict)%100 == 0:
        print(len(lei_dict), "Successes")
    try:
        lei_dict[lei] = client.get_lei(lei).legal_name
    except:
        sleep(60)
        try:
            lei_dict[lei] = client.get_lei(lei).legal_name
        except:
            sleep(5)
            try:
                lei_dict[lei] = client.get_lei(lei).legal_name
            except:
                print("Failure at",lei)
```

```python
import csv
```

```python
with open(outfname,"w",newline="",encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["LEI","Legal Name"])
    for key, value in lei_dict.items():
        writer.writerow([key,value])
    
```
